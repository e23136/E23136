# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

To find the probability density function $p(x)$ that maximizes the continuous differential entropy subject to the given constraints, we use the method of **Lagrange Multipliers**.

---

## 1. Objective Function and Constraints

The **differential entropy** $H(X)$ of a continuous random variable $X \in (-\infty, \infty)$ is defined as:
$$H(X) = -\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx$$

We maximize $H(X)$ subject to three distinct constraints:
1. **Normalization (Total Probability):** $$\int_{-\infty}^{\infty} p(x) \, dx = 1$$
2. **The Mean:** $$\int_{-\infty}^{\infty} x p(x) \, dx = \mu$$
3. **The Variance:** $$\int_{-\infty}^{\infty} (x - \mu)^2 p(x) \, dx = \sigma^2$$

---

## 2. Setting Up the Lagrangian

We define the Lagrangian functional $L(p, \lambda_0, \lambda_1, \lambda_2)$ with multipliers $\lambda_0, \lambda_1, \lambda_2$:

$$L = -\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx - \lambda_0 \left( \int_{-\infty}^{\infty} p(x) \, dx - 1 \right) - \lambda_1 \left( \int_{-\infty}^{\infty} x p(x) \, dx - \mu \right) - \lambda_2 \left( \int_{-\infty}^{\infty} (x - \mu)^2 p(x) \, dx - \sigma^2 \right)$$

Combining the integrals into a single expression:
$$L = \int_{-\infty}^{\infty} \left[ -p(x) \ln p(x) - \lambda_0 p(x) - \lambda_1 x p(x) - \lambda_2 (x - \mu)^2 p(x) \right] dx + \lambda_0 + \lambda_1 \mu + \lambda_2 \sigma^2$$

---

## 3. Maximizing the Functional

Taking the functional derivative with respect to $p(x)$ at a specific point $x$ and setting it to zero ($\frac{\delta L}{\delta p(x)} = 0$):

$$- \ln p(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2 = 0$$

Solving for $\ln p(x)$:
$$\ln p(x) = -(1 + \lambda_0) - \lambda_1 x - \lambda_2 (x - \mu)^2$$

Exponentiating both sides gives the functional form of our distribution:
$$p(x) = e^{-(1 + \lambda_0)} e^{-\lambda_1 x} e^{-\lambda_2 (x - \mu)^2}$$

---

## 4. Solving for the Lagrange Multipliers

To make solving the integrals cleaner, we can re-parameterize the linear and quadratic terms in the exponent. Let's rewrite the exponent as a complete single quadratic term centered around $\mu$:

$$p(x) = C e^{-\lambda_2 (x - \mu_0)^2}$$

By applying the Mean Constraint $\mathbb{E}[X] = \mu$, the distribution must be perfectly symmetric about $\mu$. Because a Gaussian function is symmetric only around its offset peak, it forces $\mu_0 = \mu$ and eliminates the independent linear term ($\lambda_1 = 0$).

Thus, the expression simplifies to:
$$p(x) = e^{-(1 + \lambda_0)} e^{-\lambda_2 (x - \mu)^2}$$

### Step A: Applying the Normalization Constraint
$$\int_{-\infty}^{\infty} e^{-(1 + \lambda_0)} e^{-\lambda_2 (x - \mu)^2} \, dx = 1$$
$$e^{-(1 + \lambda_0)} \int_{-\infty}^{\infty} e^{-\lambda_2 (x - \mu)^2} \, dx = 1$$

Using the standard Gaussian integral identity $\int_{-\infty}^{\infty} e^{-a(x-b)^2} \, dx = \sqrt{\frac{\pi}{a}}$:
$$e^{-(1 + \lambda_0)} \sqrt{\frac{\pi}{\lambda_2}} = 1 \implies e^{-(1 + \lambda_0)} = \sqrt{\frac{\lambda_2}{\pi}}$$

Substituting this constant back into $p(x)$:
$$p(x) = \sqrt{\frac{\lambda_2}{\pi}} e^{-\lambda_2 (x - \mu)^2}$$

### Step B: Applying the Variance Constraint
$$\int_{-\infty}^{\infty} (x - \mu)^2 p(x) \, dx = \sigma^2$$
$$\sqrt{\frac{\lambda_2}{\pi}} \int_{-\infty}^{\infty} (x - \mu)^2 e^{-\lambda_2 (x - \mu)^2} \, dx = \sigma^2$$

Using the standard Gaussian variance integral identity $\int_{-\infty}^{\infty} (x-b)^2 e^{-a(x-b)^2} \, dx = \frac{1}{2a}\sqrt{\frac{\pi}{a}}$:
$$\sqrt{\frac{\lambda_2}{\pi}} \cdot \left( \frac{1}{2\lambda_2} \sqrt{\frac{\pi}{\lambda_2}} \right) = \sigma^2$$
$$\frac{1}{2\lambda_2} = \sigma^2 \implies \lambda_2 = \frac{1}{2\sigma^2}$$

---

## 5. Final Result

Substituting $\lambda_2 = \frac{1}{2\sigma^2}$ back into our expression for $p(x)$:

$$p(x) = \sqrt{\frac{1}{2\pi\sigma^2}} e^{-\frac{(x - \mu)^2}{2\sigma^2}}$$

Which can be written compactly as:
$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

This completes the proof that the **Normal (Gaussian) Distribution** is the maximum entropy distribution given a fixed mean and variance over an infinite support interval.
