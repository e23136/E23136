```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Joint Density (Contour Plot)
t = np.linspace(-3, 3, 100)
p = np.linspace(-6, 6, 100)
T, P = np.meshgrid(t, p)
# Joint density formula: 1/(4*pi) * exp(-t^2/2 - p^2/8)
Z = (1/(4*np.pi)) * np.exp(-T**2/2 - P**2/8)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.contourf(T, P, Z, cmap='Blues')
plt.title('Joint Density of Y = (T, P)')
plt.xlabel('Temperature (X)')
plt.ylabel('Pressure')

# 2. Marginal Density of X
plt.subplot(1, 2, 2)
x = np.linspace(-4, 4, 100)
plt.plot(x, norm.pdf(x, 0, 1), color='red', lw=2)
plt.title('Marginal Density of X (Temperature)')
plt.xlabel('t')
plt.ylabel('f_X(t)')

plt.tight_layout()
plt.show()
```
